Executive Summary

Based on an analysis of 1,500 customers across nine data tables (customers, subscriptions, transactions, support tickets, and marketing campaigns), the following key findings emerged:

Key Findings:

Average revenue per customer stands at approximately 984, with the UAE, Singapore, and India generating the highest total revenue by country — indicating that high-value customers are geographically diverse rather than concentrated in a single market, pointing to strong expansion potential.

The Enterprise plan is the most valuable financially despite not being the most popular. While Professional is the most common plan by customer count, Enterprise generates the highest Monthly Recurring Revenue (572K) — meaning a small number of Enterprise customers contribute more value than a much larger group on cheaper plans.

Unexpectedly, customers with an active subscription spend slightly less on average (982) than customers without one (1007), suggesting a meaningful share of revenue comes from one-off transactions rather than recurring subscriptions — worth investigating further.

LinkedIn is by far the strongest marketing channel, driving more than double the conversions of the next-best channel (Webinar) and significantly outperforming Email and Google Ads.

Customer satisfaction is relatively low and strikingly consistent regardless of subscription status (Active/Cancelled/Paused/Expired), averaging around 3 out of 5 — indicating satisfaction is not tied to subscription status, but is low overall and needs attention.

Support ticket prioritization appears ineffective: resolution times are nearly identical across all priority levels (Critical: ~16 hours, Low: ~15.6 hours), meaning Critical tickets are not being resolved meaningfully faster than lower-priority ones.

The refund rate is notably high — roughly 25% of all transactions are refunds — a figure that warrants deeper investigation into cancellation and refund drivers.

Revenue remains stable throughout the year with no strong seasonal pattern.

Recommendations:

Revisit the support ticket prioritization system so Critical tickets are actually resolved faster than lower-priority ones.
Investigate the root causes behind the high refund rate (25%), as it may signal product quality or expectation-mismatch issues.
Increase investment in LinkedIn as a marketing channel while reducing reliance on lower-performing channels like Google Ads.
Focus retention and upsell efforts on migrating customers toward the Enterprise plan, given its outsized revenue contribution.
Follow up on the 20 "orphan" transactions (user_id 900000–900019) not linked to any known customer to rule out a data integrity issue.

## Project Summary / Data Pipeline Overview

**Data Sources (9 files):**
The project used 9 raw CSV files: `customers`, `subscriptions`, `support_tickets`, `marketing_campaigns`, `campaign_interactions`, and four quarterly transaction files (`transactions_q1` through `transactions_q4`).

**1. Cleaning (performed on each of the 9 tables individually):**
- Converted mixed-format date columns using `pd.to_datetime(..., format='mixed')`.
- Converted currency/numeric text columns (containing symbols like `$`, `USD`, commas) to proper numeric types.
- Standardized inconsistent text (extra whitespace, mixed casing) using `.str.strip()` and `.str.title()` / `.str.capitalize()`.
- Removed exact duplicate rows from every table.
- Handled missing values contextually: filled with median/mode where appropriate (e.g. `amount`, `resolution_hours`), and intentionally left as NaN where missing data has real meaning (e.g. `end_date` for active subscriptions, `satisfaction` for unrated tickets).
- Fixed invalid/sentinel values (e.g. negative `resolution_hours`, placeholder value `99999`, negative `annual_revenue`).
- Removed duplicate customer records that only differed in formatting (e.g. email casing).

**2. Combining the Quarterly Transactions (CONCAT):**
Standardized column names across `transactions_q1`–`q4`, then combined them into a single `df_transactions` table (12,000 rows) using `pd.concat`.

**3. Building Customer 360 (MERGE):**
Summarized transactions, subscriptions, support_tickets, and campaign_interactions to one row per customer using `groupby('user_id')`, then merged these summaries with the base `customers` table via left joins, producing a single `customer_360` table (1,500 rows — one per customer) with no row duplication.

**4. Analysis:**
- Calculated 10 KPIs covering revenue, subscriptions, support, and marketing performance.
- Built 8 visualizations (bar, pie, histogram, scatter, and line charts) to explore distributions, relationships, and trends.
- Answered 10 business questions using grouped and merged data to uncover actionable patterns.
- Summarized findings and recommendations in the Executive Summary above.

**Data Quality Note:**
20 transactions were found referencing customer IDs (900000–900019) not present in the `customers` table. These were excluded from `customer_360` via the left join and are flagged as likely test data or entry errors requiring follow-up.

After merging, a few columns in `customer_360` still contained nulls, and each was handled based on what the null actually meant:
- `total_amount`, `avg_amount`, and `total_monthly_fee` were null simply because some customers had no matching rows in `transactions_summary` or `subscriptions_summary` (e.g. a customer with no monthly subscription, or no transactions at all). Since the null genuinely means "zero," these were filled with `0`.
- `avg_satisfaction` and `avg_resolution_hours` were left as `NaN` intentionally. A null here means the customer either had no support tickets at all, or had tickets that were never rated — filling these with `0` would falsely imply the lowest possible satisfaction score and would distort the overall average, so they were left as missing data instead.
